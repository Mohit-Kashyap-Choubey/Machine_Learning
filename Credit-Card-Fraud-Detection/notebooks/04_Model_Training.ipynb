```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
```

```python
X_train = pd.read_csv("../dataset/X_train.csv")
X_test = pd.read_csv("../dataset/X_test.csv")

y_train = pd.read_csv("../dataset/y_train.csv").squeeze()
y_test = pd.read_csv("../dataset/y_test.csv").squeeze()
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
log_reg = LogisticRegression(
    class_weight="balanced",
    random_state=42,
    max_iter=1000
)
```

```python
log_reg.fit(X_train, y_train)
```

```python
y_pred_lr = log_reg.predict(X_test)
```

```python
y_pred_lr[:20]
```

```python
y_test[:20].values
```

```python
y_prob_lr = log_reg.predict_proba(X_test)[:, 1]
```

```python
y_prob_lr[:20]
```

```python
pd.Series(y_pred_lr).value_counts()
```

```python
pd.Series(y_pred_lr).value_counts(normalize=True) * 100
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)
```

```python
accuracy = accuracy_score(y_test, y_pred_lr)
print("Accuracy:", accuracy)
```

```python
precision = precision_score(y_test, y_pred_lr)
recall = recall_score(y_test, y_pred_lr)
f1 = f1_score(y_test, y_pred_lr)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
cm = confusion_matrix(y_test, y_pred_lr)
print(cm)
```

```python
tn, fp, fn, tp = cm.ravel()

print("True Negatives:", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives:", tp)
```

```python
print(classification_report(y_test, y_pred_lr))
```

```python
roc_auc = roc_auc_score(y_test, y_prob_lr)
print("ROC-AUC:", roc_auc)
```

```python
print("Logistic Regression Results")
print("--------------------------")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
print("ROC-AUC  :", roc_auc)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
decision_tree = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=42,
    max_depth=10
)
```

```python
decision_tree.fit(X_train, y_train)
```

```python
y_pred_dt = decision_tree.predict(X_test)
```

```python
y_prob_dt = decision_tree.predict_proba(X_test)[:, 1]
```

```python
pd.Series(y_pred_dt).value_counts()
```

```python
pd.Series(y_pred_dt).value_counts(normalize=True) * 100
```

```python
print("Tree depth:", decision_tree.get_depth())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# Calculate metrics
accuracy_dt = accuracy_score(y_test, y_pred_dt)
precision_dt = precision_score(y_test, y_pred_dt)
recall_dt = recall_score(y_test, y_pred_dt)
f1_dt = f1_score(y_test, y_pred_dt)
roc_auc_dt = roc_auc_score(y_test, y_prob_dt)

# Print results
print("Decision Tree Results")
print("---------------------")
print("Accuracy :", accuracy_dt)
print("Precision:", precision_dt)
print("Recall   :", recall_dt)
print("F1 Score :", f1_dt)
print("ROC-AUC  :", roc_auc_dt)
```

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)

print("\nConfusion Matrix:")
print(cm_dt)

tn_dt, fp_dt, fn_dt, tp_dt = cm_dt.ravel()

print("\nTrue Negatives :", tn_dt)
print("False Positives:", fp_dt)
print("False Negatives:", fn_dt)
print("True Positives :", tp_dt)
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

```python
print("\nFeature Importances:")
print(
    pd.Series(
        decision_tree.feature_importances_,
        index=X_train.columns
    ).sort_values(ascending=False)
)
```
