# Conclusion

This project focused on detecting fraudulent credit card transactions using machine learning.

The dataset contained approximately 1 million transactions with seven input features. Exploratory analysis showed that fraudulent transactions differed significantly from legitimate transactions in features such as ratio to median purchase price, transaction distance and online ordering.

Three classification models were evaluated:

- Logistic Regression
- Decision Tree
- Random Forest

Logistic Regression achieved a strong recall of approximately 90.55%, but its precision was considerably lower. The Decision Tree and Random Forest models performed substantially better on this dataset, achieving almost perfect precision, recall and F1 scores.

The Decision Tree achieved the highest F1 score among the three models, while Random Forest produced a very similar result. Tree-based models also showed that `ratio_to_median_purchase_price` was the most influential feature.

Although the models achieved near-perfect performance, these results should be interpreted carefully. The dataset contains very strong patterns separating fraudulent and legitimate transactions, so the performance may not represent what would be achieved on real-world financial transaction data.

Overall, the project demonstrates the complete machine learning workflow from data exploration and preprocessing to model training, evaluation and threshold analysis.

# Limitations and Future Improvements

- The dataset may not fully represent real-world credit card transactions.
- The extremely high model performance may indicate that the dataset contains unusually strong patterns.
- Real-world fraud detection would require continuous monitoring and model retraining.
- Threshold selection would depend on the business cost of false positives and false negatives.
- Future work could involve testing the models on a more realistic or external dataset.
- Model interpretability and deployment could also be explored in future versions.
